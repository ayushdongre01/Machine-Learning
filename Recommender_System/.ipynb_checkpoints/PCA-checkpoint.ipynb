```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## Sample Dataset

```python
X = np.array([
    [2.5, 2.4],
    [0.5, 0.7],
    [2.2, 2.9],
    [1.9, 2.2],
    [3.1, 3.0],
    [2.3, 2.7],
    [2.0, 1.6],
    [1.0, 1.1],
    [1.5, 1.6],
    [1.1, 0.9]
])

df = pd.DataFrame(X, columns=["Feature1", "Feature2"])
print(df.head())
```

#### Standardize the data

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

#### Apply PCA

```python
pca = PCA(n_components = 1)
X_pca = pca.fit_transform(X_scaled)
```

#### Explained Variance

```python
print("Explained Variance Ratio:", pca.explained_variance_ratio_)
print("Total Variance Retained:", np.sum(pca.explained_variance_ratio_))
```

## Visualization

```python
plt.scatter(X_scaled[:,0], X_scaled[:,1], color='red')
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Original Scaled Data")
plt.show()
```

#### PCA Reduced Data (1D -> Projected)

```python
plt.scatter(X_pca[:,0], np.zeros(len(X_pca)), color='blue')
plt.xlabel("Principal Component 1")
plt.title("PCA Reduced Data (1D)")
plt.show()
```

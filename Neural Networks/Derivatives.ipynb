```python
from sympy import symbols, diff
```

## Informal Definition of derivative

```python
J = (3)**2
J_epsilon = (3 + 0.001)**2
k = (J_epsilon - J)/0.001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k:0.6f}")
```

```python
J = (3)**2
J_epsilon = (3 + 0.000000001)**2
k = (J_epsilon - J)/0.000000001
print(f"J = {J}, J_epsilon = {J_epsilon}, dJ_dw ~= k = {k} ")
```

## Finding symbolic derivatives

##  $J = w^2$

```python
J, w = symbols('J, w')
```

```python
J = w**2
J
```

```python
dJ_dw = diff(J, w)
dJ_dw
```

```python
dJ_dw.subs([(w, 2)])
```

```python
dJ_dw.subs([(w, 3)])
```

```python
dJ_dw.subs([(w, -3)])
```

##  $J = 2w$

```python
w, J = symbols('w, J')
```

```python
J = 2*w
J
```

```python
dJ_dw = diff(J, w)
dJ_dw
```

```python
dJ_dw.subs([(w, -3)])
```

```python
J, w = symbols('J, w')
```
